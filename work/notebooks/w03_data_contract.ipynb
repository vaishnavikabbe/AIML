## 2. Fields: feature / label / context / excluded

### Fields: feature / label / context / excluded

**Features:** I will use `search_volume`, `impressions_90d`, and `word_count` as candidate features. These fields describe measurable search-performance and content characteristics of each page.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### Data limits

This dataset has several limitations that should be considered when interpreting the results.

First, the dataset is anonymized and represents only the available data in the starter release. It should not be treated as a complete representation of every search situation.

Second, some performance fields are measured over fixed windows such as 90 days. These windows may overlap, so observations should not automatically be treated as independent historical periods.

Third, the dataset can show observed relationships between page signals and search performance, but it cannot prove that one factor causes another.

Finally, the analysis cannot explain or predict Google's ranking algorithm. The results will be used as observed, measured, directional, and decision-support evidence for prioritizing pages for review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ML-04 Section 4: Check data limitations that can be measured

print("=== DATA LIMIT CHECK ===")

print("Total rows:", len(df))
print("Total columns:", len(df.columns))

print("\nMissing values in main fields:")

for col in [
    "search_volume",
    "impressions_90d",
    "word_count",
    "trend_direction"
]:
    print(f"{col}: {df[col].isna().sum()}")

print("\nObserved trend categories:")
print(df["trend_direction"].dropna().unique())

print("\n90-day performance fields are fixed-window measurements.")
print("They should not be interpreted as independent observations across time.")
```
